## 2. EDA â€” Message Length Distribution & Top Bigrams

````python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# --- Message length histogram ---
axes[0].hist(df['msg_length'], bins=40, color='mediumslateblue', edgecolor='white')
axes[0].set_title('Message Length Distribution')
axes[0].set_xlabel('Characters')
axes[0].set_ylabel('Count')
axes[0].axvline(df['msg_length'].median(), color='red',
                linestyle='--', label=f'Median={df["msg_length"].median():.0f}')
axes[0].legend()

# --- Top-20 bigrams ---
tfidf = TfidfVectorizer(ngram_range=(2, 2), max_features=20,
                         stop_words='english', min_df=2)
tfidf.fit(df['content'].fillna(''))
bigrams = tfidf.get_feature_names_out()
scores  = tfidf.transform(df['content'].fillna('')).toarray().sum(axis=0)
order   = np.argsort(scores)[::-1]
axes[1].barh(bigrams[order][::-1], scores[order][::-1], color='coral')
axes[1].set_title('Top-20 Bigrams (TF-IDF)')
axes[1].set_xlabel('Aggregate TF-IDF score')

plt.tight_layout()
plt.savefig('../models/02_eda.png', dpi=100)
plt.show()
````

## 3. Embedding

`all-MiniLM-L6-v2` (sentence-transformers) maps each message to a 384-dimensional dense vector.
The model is lightweight (~80 MB) and runs on CPU in production.

````python
from sentence_transformers import SentenceTransformer
import numpy as np

encoder = SentenceTransformer('all-MiniLM-L6-v2')

messages = df['content'].fillna('').tolist()
embeddings = encoder.encode(messages, batch_size=64,
                             show_progress_bar=True, normalize_embeddings=True)

print('Embedding matrix shape:', embeddings.shape)
print('Sample embedding (first 8 dims):', embeddings[0, :8].round(4))

# Persist for reuse in steps below
np.save('../data/processed/chat_embeddings.npy', embeddings)
````

## 4. Baseline â€” KMeans k=8

A flat KMeans with k=8 clusters serves as the baseline. Silhouette score on the full embedding matrix quantifies within-cluster cohesion.

````python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import numpy as np

embeddings = np.load('../data/processed/chat_embeddings.npy')

km = KMeans(n_clusters=8, random_state=42, n_init='auto')
km_labels = km.fit_predict(embeddings)

km_sil = silhouette_score(embeddings, km_labels, sample_size=min(2000, len(km_labels)))
print(f'KMeans  n_clusters = 8')
print(f'KMeans  silhouette = {km_sil:.4f}')
print(f'KMeans  noise %    = 0.00  (KMeans assigns every point)')
````

## 5. sklearn.cluster.OPTICS Clustering

sklearn.cluster.OPTICS discovers variable-density clusters without requiring k upfront. Points not belonging to any dense region are labelled âˆ’1 (noise). A lower noise % alongside a higher silhouette score indicates a better-quality clustering.

````python
import sklearn.cluster.OPTICS
import numpy as np
from sklearn.metrics import silhouette_score

embeddings = np.load('../data/processed/chat_embeddings.npy')

clusterer = sklearn.cluster.OPTICS.sklearn.cluster.OPTICS(
    min_cluster_size=5,
    min_samples=3,
    metric='euclidean',
    cluster_selection_method='eom'
)
hdb_labels = clusterer.fit_predict(embeddings)

n_clusters  = len(set(hdb_labels)) - (1 if -1 in hdb_labels else 0)
noise_pct   = (hdb_labels == -1).mean() * 100

# Silhouette only on non-noise points (need â‰¥2 clusters)
mask = hdb_labels != -1
if n_clusters >= 2 and mask.sum() > n_clusters:
    hdb_sil = silhouette_score(embeddings[mask], hdb_labels[mask],
                                sample_size=min(2000, mask.sum()))
else:
    hdb_sil = float('nan')

print(f'sklearn.cluster.OPTICS  n_clusters = {n_clusters}')
print(f'sklearn.cluster.OPTICS  noise %    = {noise_pct:.1f}%')
print(f'sklearn.cluster.OPTICS  silhouette = {hdb_sil:.4f}')
````

## 6. Comparison Table

````python
import pandas as pd

results = pd.DataFrame([
    {
        'Method':     'KMeans (k=8)',
        'n_clusters': 8,
        'silhouette': round(km_sil, 4),
        'noise_%':    '0.0%'
    },
    {
        'Method':     'sklearn.cluster.OPTICS',
        'n_clusters': n_clusters,
        'silhouette': round(hdb_sil, 4) if hdb_sil == hdb_sil else 'N/A',
        'noise_%':    f'{noise_pct:.1f}%'
    },
]).set_index('Method')

print(results.to_string())
print('\nHigher silhouette = tighter, better-separated clusters (range: -1 to 1).')
````

## 7. Limitations

- Small synthetic corpus: cluster stability (bootstrapped Jaccard) is not yet measured.
- Embeddings computed with `all-MiniLM-L6-v2`; domain-specific fine-tuning may improve semantic grouping for STEM content.
- UMAP / t-SNE projection is omitted here; a 2D scatter plot would aid cluster interpretation.
- sklearn.cluster.OPTICS `min_cluster_size` is a sensitive hyperparameter; grid search over (3â€“15) is recommended with more data.
- Noise points (-1 labels) are excluded from silhouette calculation, which may overstate cluster quality when noise % is high.
- Cluster labels are not automatically named; a post-hoc LLM summarisation step is needed in production.

## 8. Production Usage

Use `get_user_clusters(user_id)` from `analytics.ml.confusion_clusters`.

The function:
1. Fetches recent user messages from the database.
2. Embeds them with the cached `SentenceTransformer` singleton.
3. Assigns each message to the nearest persisted sklearn.cluster.OPTICS centroid (or `-1` for novel confusion).
4. Returns a ranked list of `{cluster_id, label, message_count, example_messages}` dicts.
